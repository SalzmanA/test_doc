 * setf assign local dof index to each dof of the field
 * foo sum local dof of a field

```python
%%px
from dolfinx import mesh
from dolfinx import fem
import bar


domain=mesh.create_interval(MPI.COMM_WORLD,5,[0.,1.])
space=fem.functionspace(domain,domain.ufl_domain().ufl_coordinate_element())
field=fem.Function(space)
bar.setf(field)
print(field.x.array)
print(bar.foo(field._cpp_object))
```
